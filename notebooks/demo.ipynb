# Vortex-AI Demo

Load prepared NIFTY-50 artifacts, inspect regimes, and visualize returns and empirical graphs.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / "data" / "raw").exists():
    ROOT = ROOT.parent
RAW = ROOT / "data" / "raw"
returns = np.load(RAW / "returns.npy")
regimes = np.load(RAW / "regimes.npy")
windows = np.load(RAW / "windows.npy")
adjacency = np.load(RAW / "adj_matrices.npy")
print("returns:", returns.shape)
print("windows:", windows.shape)
print("adjacency:", adjacency.shape)
print("crisis days:", int(regimes.sum()), "/", len(regimes))
```

```python
cumulative = np.cumprod(1.0 + returns.mean(axis=1))
fig, axis = plt.subplots(figsize=(12, 4))
axis.plot(cumulative, label="Equal-weighted portfolio")
axis.set_title("Cumulative Returns")
axis.set_xlabel("Trading day")
axis.legend()
plt.show()
```

```python
normal_index = int(np.flatnonzero(np.load(RAW / "window_regimes.npy") == 0)[0])
crisis_index = int(np.flatnonzero(np.load(RAW / "window_regimes.npy") == 1)[0])
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].imshow(adjacency[normal_index], cmap="Blues")
axes[0].set_title("Normal adjacency")
axes[1].imshow(adjacency[crisis_index], cmap="Reds")
axes[1].set_title("Crisis adjacency")
for axis in axes:
    axis.set_xticks([])
    axis.set_yticks([])
plt.tight_layout()
plt.show()
```
